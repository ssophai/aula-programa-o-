Funções em python

```python
def saudacao():
  print('Olá, estou dentro da função')

def somar():
  print(5 + 8)

saudacao() #para que a função seja printada, é necessário chamar, ex.: saudacao()
somar()
```

```python
def quadrado(a):
  return a**2

num = int(input('Digite um número: '))
resultado = quadrado(num)
print(resultado)
```

```python
#Padrão (default)
def exibir_horario_aulas(disciplina, horario=9):
    print(f'{disciplina} - {horario} horas.')

exibir_horario_aulas('Programação')
exibir_horario_aulas('Modelagem de Software', 15)
```

```python
def exibir_produto(nome, preco):
  print(f'{nome} : R${preco}')

#Argumentos posicionais
exibir_produto('doce de leite', 5.50)
exibir_produto(65, 'Bermuda')

#Argumentos Nomeados
exibir_produto(preco=100, nome='Camiseta')
```

Forçar argumentos nomeados com *

```python
def exibir_jogo(*, nome_jogo, ano_lancamento):
  print(f'{nome_jogo}, lançado em {ano_lancamento}')

#apenas um parametro deve ser nomeado
def exibir_produto(nome,*, preco):
  print(f'{nome} : R${preco}')

exibir_jogo(nome_jogo='God of War', ano_lancamento=2018)
exibir_produto('Camiseta', preco=100)
```

```python
def gerar_personagem_game(nome, altura=1.75,*, forca, arma):
  print(f'Meu nome é {nome}, tenho {altura}, minha força equivale a {forca} e uso {arma} como arma.')

gerar_personagem_game('Antonio', 1.60, forca=50, arma='revolver')
```

*ARGS

```python
def exemplo(*args):
  print(args)

exemplo(1, 2, 3, 4, 5)
```

```python
def somanotas(*args):
  return sum(args)

print(somanotas(7,8,3,10,5.5,2,8.5))
```

```python
def somar_nums(*valores, a):
  print(valores)
  for valor in valores:
     a += valor
  return a

print(somar_nums(5,6,9,10, a=10))
```

kwargs

```python
def exemplo(**kwargs):
  print(kwargs)

exemplo(a=1, b=2)
```

```python
def concatenar(**plavras):
  frase = ''
  for palavra in plavras.values():
    frase += palavra
  print(frase)

concatenar(a='Eu ', b='é ', c='Hequer')
```

```python
#Exemplo

def calcula(nome, *args, **kwargs):
  print(nome)
  print(args)
  print(kwargs)
  for arg in args:
    print(arg)
  for kwarg in kwargs.values():
    print(kwarg)

calcula('José',5,6,72,1,a=4,b=6,c=8)
```

```python
def somar_valores(*args,**kwargs):
  soma = 0
  for valor in args:
    soma += valor
  for valor in kwargs.values():
    soma += valor
  return soma

somar_valores(5,6,72,1,a=4,b=6,c=8)
```

```python
def exibir_perfil(*hobbies,nome, idade=0,cidade, **dados_pessoais):
  print(f'Nome: {nome}')
  print(f'Idade: {idade}')
  print(f'Hobbies:')
  for hobby in hobbies:
    print(f'- {hobby}')
  print(f'Dados Pessoais: {dados_pessoais}')
  for dado in dados_pessoais.keys():
    print(f'{dado}: {dados_pessoais[dado]}')

exibir_perfil('programar', 'ler', 'cinema', 'tocar teclado',nome='Antonio', idade=16, cidade='João Pessoa')
```

```python
def multiplicar_valores(*args):
  multiplica = 1
  for valor in args:
    multiplica *= valor
  return multiplica

multiplicar_valores(5,6,72)
```

datetime

```python
from datetime import date, datetime

data_atual = date.today()
datahora_atual = datetime.now()
print(data_atual)
print(datahora_atual)
```

```python
#Data formatada

data_formatada = data_atual.strftime('%d/%m/%Y')
data_completa = datahora_atual.strftime('%d/%m/%Y %H:%M:%S')
print(data_formatada)
print(data_completa)
```
